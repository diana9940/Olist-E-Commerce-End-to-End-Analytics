```python
import urllib.parse
from sqlalchemy import create_engine

params = urllib.parse.quote_plus(
    "DRIVER={ODBC Driver 17 for SQL Server};"
    "SERVER=localhost\\SQLEXPRESS;"
    "DATABASE=Olist;"  
    "Trusted_Connection=yes;"
)

engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")

orderData.to_sql('orderData', con=engine, if_exists='replace', index=False)
```

```python
order_item.to_sql('order_item', con=engine, if_exists='replace', index=False)
payments.to_sql('payments', con=engine, if_exists='replace', index=False)
translation.to_sql('translation', con=engine, if_exists='replace', index=False)
reviews.to_sql('reviews', con=engine, if_exists='replace', index=False)
sellers.to_sql('sellers', con=engine, if_exists='replace', index=False)
geolocation.to_sql('geolocation', con=engine, if_exists='replace', index=False)
customer.to_sql('customer', con=engine, if_exists='replace', index=False)
products.to_sql('products', con=engine, if_exists='replace', index=False)
```
